### CALENDAR

```python
calendar
```

```python
# Verificar si hay valores nulos
if calendar.isnull().any().any():
    print("El DataFrame tiene valores faltantes en:\n",calendar.isnull().sum())
else:
    print("El DataFrame no tiene valores faltantes.")
```

```python
# Verificar si hay duplicados
print(calendar.duplicated().sum())
```

```python
calendar.info()
```

```python
# Asegurar que 'YEAR' sea un número de 4 dígitos
calendar['YEAR'] = calendar['YEAR'].astype(str)
calendar['YEAR'] = calendar['YEAR'].apply(lambda x: x if len(x) == 4 else None)
# Convertir a número
calendar['YEAR'] = pd.to_numeric(calendar['YEAR'], errors='coerce')
calendar
```

```python
# Asegurar que el mes esté entre 1 y 12
calendar['MONTH'] = calendar['MONTH'].apply(lambda x: x if 1 <= x <= 12 else None)
calendar
```

```python
# Asegurar que el número de semana esté entre 1 y 53
calendar['WEEK_NUMBER'] = calendar['WEEK_NUMBER'].apply(lambda x: x if 1 <= x < 53 else None)
calendar
```

```python
# Extraer el número de semana y año de la columna 'WEEK'
calendar_ext = calendar['WEEK_NUMBER_EXTRACTED'] = calendar['WEEK'].apply(lambda x: int(x.split('-')[0]))
calendar_year = calendar['YEAR_EXTRACTED'] = calendar['WEEK'].apply(lambda x: int(x.split('-')[1]))

# Verificar si la fecha corresponde a la semana indicada
calendar['DATE_WEEK'] = calendar['DATE'].dt.isocalendar().week

# Comprobar si el número de semana extraído coincide con el de la columna 'WEEK_NUMBER'
calendar['WEEK_CHECK'] = calendar['WEEK_NUMBER'] == calendar['DATE_WEEK']

# Verificar si hay inconsistencias
calendar[calendar['WEEK_CHECK'] == False]
```
