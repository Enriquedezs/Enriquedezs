### Visualizar la distribución de ventas:

Crea gráficos de distribución (histogramas o boxplots) para visualizar la distribución de ventas por diferentes categorías. Esto ayudará a identificar patrones de ventas y posibles outliers.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid") # Ajustes generales de estilo
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df_consolidado, x="SEGMENT", y="TOTAL_UNIT_AVG_WEEKLY_SALES")
plt.title("Boxplot de valor de ventas por segmento")
plt.xlabel("Segmento")
plt.ylabel("Valor total de ventas")
plt.xticks(rotation=45)
plt.show()
```

##### CONCLUSIÓNES:
1. Algunos segmentos concentran más ventas, mientras que otros tienen menor volumen.
2. "Bleach" domina de forma abrumadora: su media supera $$ 100 y presenta outliers de hasta $12000.
3. “Liquid & gel” y “powder” son los siguientes, con picos alrededor de 1500 hasta 2500 .
   
$Conclusión$: El segmento de blanqueadores es, con diferencia, el motor principal de ingresos; los demás son nichos mucho menores. El negocio gira principalmente en torno a los productos blanqueadores. Deben ser prioridad total en producción, marketing y distribución.

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df_consolidado, x="REGION", y="TOTAL_VALUE_SALES")
plt.title("Boxplot de valor de ventas por Región")
plt.xlabel("Región")
plt.ylabel("Valor total de ventas")
plt.xticks(rotation=45)
plt.show()
```

##### CONCLUSIONES
1. Entre las 6 areas, Area 2 y Area 5 son las que muestran los rangos y outliers mas altos.
2. Las 6 areas se mueven en niveles parecidos; no hay una que aplaste a las demas.

$Conclución:$ El volumen esta bastante repartido entre las areas. Area 2 y Area 5 son las mas fuertes, pero por poco margen respecto al resto.

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df_consolidado, x="FORMAT", y="TOTAL_VALUE_SALES")
plt.title("Boxplot de valor de ventas por Formato")
plt.xlabel("Format")
plt.ylabel("Valor total de ventas")
plt.xticks(rotation=45)
plt.show()
```

##### CONCLUCIÓN
1. Formato “líquido” sobresale con distribuciones que alcanzan 12,000; “gel” es segundo.
2. “Polvo”, “barra” y “toallas” mantienen valores mucho más modestos.

$Conclusión$ : La presentación líquida es claramente la preferida — conviene priorizarla en inventario y campañas. El formato líquido es el más rentable y demandado. Los esfuerzos deben orientarse a proteger su cadena de suministro.

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df_consolidado, x="MANUFACTURER", y="TOTAL_VALUE_SALES")
plt.title("Boxplot de valor de ventas por Fabricante")
plt.xlabel("Fabricante")
plt.ylabel("Valor total de ventas")
plt.xticks(rotation=80)
plt.show()
```

### Analizar la tendencia de ventas en el tiempo:

Genera gráficos de líneas que muestren las tendencias de ventas a lo largo del tiempo. Filtra las ventas por diferentes productos, regiones o segmentos para analizar cómo han cambiado las ventas a lo largo del tiempo.

```python
# Separar 'WEEK' en dos columnas: semana y año
df_consolidado[["SEMANA", "AÑO"]] = df_consolidado["WEEK"].str.split("-", expand=True)
df_consolidado["SEMANA"] = df_consolidado["SEMANA"].astype(int)
df_consolidado["AÑO"] = df_consolidado["AÑO"].astype(int) + 2000  # Asegura que 22 sea 2022

# Crear columna de fecha a partir del año y semana (primer día de la semana)
df_consolidado["WEEK_DATE"] = pd.to_datetime(df_consolidado["AÑO"].astype(str) + "-W" + df_consolidado["SEMANA"].astype(str) + "-1", format="%G-W%V-%u")

```

```python
df_consolidado["WEEK_DATE"]
```

### GRÁFICO DE LINEAS

```python
#TENDENCIA GENERAL DE VENTAS SEMANALES
# Agrupar por semana y sumar ventas
ventas_total = df_consolidado.groupby("WEEK_DATE")["TOTAL_VALUE_SALES"].sum().reset_index() #convierte el resultado en un DataFrame normal

plt.figure(figsize=(12,6))
sns.lineplot(data=ventas_total, x="WEEK_DATE", y="TOTAL_VALUE_SALES")
plt.title("Tendencia general de ventas semanales")
plt.xlabel("Semana")
plt.ylabel("Ventas totales ($)")
plt.xticks(rotation=45) 
plt.tight_layout() #ajusta el espacio para que no se corte nada
plt.show()
```

##### CONCLUCIONES
1. Se observa tendencia ascendente suave de ~120 k  a ~150 k por semana.
2. Picos estacionales pronunciados (junio‑2022 y junio‑2023) con récord de casi 200 k.

$Conclución:$ El negocio crece gradualmente, con picos estacionales que conviene anticipar para ajustar stock. Se deben planear acciones especiales antes de los meses pico (campañas, refuerzo de stock, promociones).

```python
# Agrupar por semana y región
ventas_region = df_consolidado.groupby(["WEEK_DATE", "REGION"])["TOTAL_VALUE_SALES"].sum().reset_index()#Agrupa por semana Y por región, así puedes ver cuánto se vendió por región cada semana.

plt.figure(figsize=(14,7))
sns.lineplot(data=ventas_region, x="WEEK_DATE", y="TOTAL_VALUE_SALES", hue="REGION") #Dibuja una línea por cada región. hue="REGION" indica que cada región tendrá una línea de diferente color.
plt.title("Tendencia de ventas semanales por región") # Título del gráfico
plt.xlabel("Semana") # Nombre eje x
plt.ylabel("Ventas ($)") # Nombre eje y
plt.xticks(rotation=45) # Gira las etiquetas
plt.legend(title="Región", bbox_to_anchor=(1.05, 1), loc='upper left')#coloca la imagen de la leyenda fuera del gráfico, justo a la derecha y alinea la esquina superior izquierda de la leyenda en esa posición
plt.tight_layout() #Ajusta automáticamente los márgenes para que no se corte nada del gráfico (útil cuando hay muchas etiquetas).
plt.show()
```

##### CONCLUCIONES
1. Las 6 areas siguen patrones paralelos a lo largo del tiempo (suben y bajan casi al mismo tiempo).
2. Area 2 y Area 5 van consistentemente arriba; Area 4 es la mas baja.

$Conclusión:$ Las campañas nacionales pegan parejo en todas las regiones, con Area 2 y Area 5 liderando de forma estable.

```python
# Agrupar por semana y segmento
ventas_segmento = df_consolidado.groupby(["WEEK_DATE", "SEGMENT"])["TOTAL_VALUE_SALES"].sum().reset_index()

plt.figure(figsize=(14,7))
sns.lineplot(data=ventas_segmento, x="WEEK_DATE", y="TOTAL_VALUE_SALES", hue="SEGMENT")
plt.title("Tendencia de ventas semanales por segmento")
plt.xlabel("Semana")
plt.ylabel("Ventas ($)")
plt.xticks(rotation=45)
plt.legend(title="Segmento", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

##### CONCLUSIONES

1. Bleach oscila entre 80 k y 135 k semanales; es el claro “driver”.
2. “Liquid & gel” ronda 15 k –25 k y crece ligeramente.

$Conclusión:$ Dirigir promociones hacia “bleach” potencia los ingresos globales; diversificar en “liquid & gel” ayuda a suavizar la dependencia.

```python
# Obtener los 5 productos con más ventas totales
top5 = df_consolidado.groupby("ITEM_DESCRIPTION")["TOTAL_VALUE_SALES"].sum().nlargest(5).index

# Filtrar solo esos productos
df_top5 = df_consolidado[df_consolidado["ITEM_DESCRIPTION"].isin(top5)]

# Agrupar por semana y producto
ventas_top5 = df_top5.groupby(["WEEK_DATE", "ITEM_DESCRIPTION"])["TOTAL_VALUE_SALES"].sum().reset_index()

# Graficar
plt.figure(figsize=(14,7))
sns.lineplot(data=ventas_top5, x="WEEK_DATE", y="TOTAL_VALUE_SALES", hue="ITEM_DESCRIPTION")
plt.title("Tendencia de ventas semanales - Top 5 productos más vendidos")
plt.xlabel("Semana")
plt.ylabel("Ventas ($)")
plt.xticks(rotation=45)
plt.legend(title="Producto", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

##### CONCLUCIONES
1. Todos los SKU del Top 5 muestran crecimiento y picos alineados con las fechas señaladas en el grafico de ventas semanales.
2. El SKU “cloralex bot. 3750 ml” es el más volátil, rozando 25 k $/semana en máximos.
3. Cloralex bot. 2000 ml es el segundo más volátil.

$Conclución:$ Los líderes del portafolio responden bien a la estacionalidad y deberían recibir prioridad logística en esos momentos.

### Explorar la relación entre diferentes variables:
Utiliza gráficos de dispersión (scatter plots) para investigar relaciones entre variables clave. Esto puede ayudar a identificar correlaciones significativas entre diferentes variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.scatterplot(data=df_consolidado, x="TOTAL_UNIT_SALES", y="TOTAL_VALUE_SALES")
plt.title("Relación entre unidades vendidas y valor total de ventas")
plt.xlabel("Unidades vendidas")
plt.ylabel("Valor total de ventas ($)")
plt.tight_layout()
plt.show()
```

##### CONCLUCIONES

1. Correlación Positiva Clara: Existe una correlación positiva fuerte y evidente entre el número de unidades vendidas y el valor total de las ventas. A medida que la cantidad de unidades vendidas aumenta, el valor total de las ventas también tiende a aumentar de manera consistente
2. Patrón Lineal General: La mayoría de los puntos de datos se agrupan a lo largo de una línea ascendente, lo que sugiere una relación aproximadamente lineal entre las dos variables. 
3. Presencia de Grupos o Patrones Discretos: Observamos que los puntos no forman una única línea perfectamente continua, sino que parecen agruparse en bandas o clústeres. Por ejemplo, hay una concentración densa de puntos para valores de unidades vendidas más bajos y luego la dispersión se abre en niveles más altos, mostrando quizás diferentes rangos de precios unitarios o tipos de productos. 
4. Variabilidad en Precios Unitarios (Implícita): La dispersión de los puntos a medida que aumenta el número de unidades vendidas sugiere que el precio por unidad no es constante para todas las transacciones o productos.
5. Outliers (Posibles): Aunque no son extremadamente prominentes, hay algunos puntos que se desvían de la tendencia principal, especialmente en los rangos superiores de unidades vendidas. 

```python
plt.figure(figsize=(12,6))
sns.scatterplot(
    data=df_consolidado,
    x="TOTAL_UNIT_SALES",
    y="TOTAL_VALUE_SALES",
    hue="SEGMENT",
    alpha=0.7
)
plt.title("Unidades vs. Valor de ventas, coloreado por segmento")
plt.xlabel("Unidades vendidas")
plt.ylabel("Valor total de ventas ($)")
plt.legend(title="Segmento", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

##### CONCLUCIONES
1. Los puntos violetas (bleach) cubren todas las bandas de precio y las mayores cantidades.
2. Otros segmentos se agrupan en zonas de menor valor.

$Conclución:$ Bleach domina no solo en facturación total, sino en todos los rangos de precio‑volumen.
